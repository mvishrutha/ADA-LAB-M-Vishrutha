```python
import heapq

class PuzzleState:
    def __init__(self, board, parent=None, move="", g=0):
        self.board = board  # 1D tuple of size 9 representing the 3x3 grid
        self.parent = parent
        self.move = move
        self.g = g
        self.h = self.calculate_misplaced_tiles()
        self.f = self.g + self.h

    def __lt__(self, other):
        return self.f < other.f

    def calculate_misplaced_tiles(self):
        """
        Heuristic function: Counts the number of misplaced tiles.
        We compare each tile (excluding 0) with its position in the goal state.
        """
        misplaced = 0
        goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)
        for i in range(9):
            if self.board[i] != 0 and self.board[i] != goal[i]:
                misplaced += 1
        return misplaced

    def get_neighbors(self):
        neighbors = []
        zero_idx = self.board.index(0)
        r, c = zero_idx // 3, zero_idx % 3

        moves = {
            "UP": (-1, 0),
            "DOWN": (1, 0),
            "LEFT": (0, -1),
            "RIGHT": (0, 1)
        }

        for direction, (dr, dc) in moves.items():
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                neighbor_idx = nr * 3 + nc
                new_board = list(self.board)
                # Swap the blank tile (0) with the neighbor tile
                new_board[zero_idx], new_board[neighbor_idx] = new_board[neighbor_idx], new_board[zero_idx]
                neighbors.append(PuzzleState(tuple(new_board), self, direction, self.g + 1))
        return neighbors

def solve_a_star(start_board, goal_board):
    start_state = PuzzleState(start_board)
    open_set = []
    heapq.heappush(open_set, start_state)
    closed_set = set()

    while open_set:
        current_state = heapq.heappop(open_set)

        if current_state.board == goal_board:
            path = []
            curr = current_state
            while curr:
                path.append(curr)
                curr = curr.parent
            return path[::-1]

        closed_set.add(current_state.board)

        for neighbor in current_state.get_neighbors():
            if neighbor.board in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None

def print_board(board):
    for i in range(0, 9, 3):
        print(list(board[i:i+3]))
    print()

def get_user_board(state_name):
    while True:
        try:
            print(f"\nEnter the {state_name} (row by row, use 0 for empty space):")
            row1 = list(map(int, input("Row 1 (e.g. 1 2 3): ").strip().split()))
            row2 = list(map(int, input("Row 2 (e.g. 0 4 6): ").strip().split()))
            row3 = list(map(int, input("Row 3 (e.g. 7 5 8): ").strip().split()))

            board = tuple(row1 + row2 + row3)
            if len(board) == 9 and set(board) == set(range(9)):
                return board
            print("Invalid grid. Please ensure you enter exactly 3 numbers per row containing unique digits from 0 to 8.")
        except ValueError:
            print("Please enter numbers only.")

# User Inputs row-wise
start_layout = get_user_board("Start State")
goal_layout = get_user_board("Goal State")

solution_path = solve_a_star(start_layout, goal_layout)

if solution_path:
    print(f"\nSolvable in {len(solution_path) - 1} moves!\n")
    for idx, state in enumerate(solution_path):
        if idx == 0:
            print("Initial State:")
        else:
            print(f"Step {idx}: Move {state.move}")
        print_board(state.board)
else:
    print("No solution found.")
```
